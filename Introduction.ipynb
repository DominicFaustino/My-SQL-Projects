## If Statement

if <condition>:
    <code>
elif <condition>:
    <code>
...
else:
    <code>

```python
age = 12

if age>= 18:
    print("You're an Adult")
elif age>= 13:
    print("You're a Teenager")
elif age<= 12:
    print("You're a Kid")
else:
    print("Boomer")
```

# Loop

for <variable> in <iterable>:
    <code>

```python
countries
```

```python
for country in (countries):
    print(country)
```

```python
for i, country in enumerate (countries):
    print(i)
    print(country)
```

```python
my_data = {'name':'bean', 'age':22, 'height':5.11 }
my_data
```

```python
for key, value in my_data.items():
    print(key)
    print(value)
```
